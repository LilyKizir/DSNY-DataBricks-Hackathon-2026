```python
# Define catalog and desired schema name
catalog_name = "the_data_masons"
schema_landing = "landing"
schema_bronze = "bronze"
schema_silver = "silver"
schema_gold = "gold"
volume_landing = "raw_test"

# Remove old schemas and volumes
spark.sql(f"DROP SCHEMA IF EXISTS {catalog_name}.{schema_landing} CASCADE")
spark.sql(f"DROP SCHEMA IF EXISTS {catalog_name}.{schema_bronze} CASCADE")
spark.sql(f"DROP SCHEMA IF EXISTS {catalog_name}.{schema_silver} CASCADE")
spark.sql(f"DROP SCHEMA IF EXISTS {catalog_name}.{schema_gold} CASCADE")


# Execute the SQL DDL (Data Definition Language) command
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog_name}.{schema_bronze}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog_name}.{schema_silver}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog_name}.{schema_gold}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog_name}.{schema_landing}")
spark.sql(f"CREATE VOLUME IF NOT EXISTS {catalog_name}.{schema_landing}.{volume_landing}")
```
